```python
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
data=pd.read_csv(r"E:\python30Day\pandas\sales_prediction\myworld\sales_prediction.csv")
```

```python
data.drop_duplicates()
```

```python
data["Order Date"]=pd.to_datetime(data["Order Date"])
```

```python
data.info()
```

```python
data['Month'] = data['Order Date'].dt.month
```

```python
data['Year'] = data['Order Date'].dt.year
```

```python
data.head(5)
```

```python
monthly_sales = data.groupby(['Year' , 'Month' ]) ['Sales'].sum().reset_index()
```

```python
print(monthly_sales)
```

```python
print('\nTotal sales:', data['Sales'].sum())
```

```python
data.to_csv('clean_data.csv',index=False)
```

```python
X = monthly_sales[['Month' , 'Year']]
Y = monthly_sales['Sales']

model = LinearRegression()
model.fit(X,Y)

future = pd.DataFrame({
    'Month' :[1,2,3,4,5,6,7],
    'Year' :[2026]*7
})
predictions = model.predict(future)
future['Predicted_Sales'] = predictions

print("\nFuture Sales Prediction")
print(future)
```

```python
future.to_csv("sales_forecast.csv", index=False)
```
